```python
def average(array):
    sums = sum(set(array))
    avg = sums / len(set(array))
    return avg
```

```python
array_len, set_len  = list(map(int, input().split()))
array = input().split()
set_a = set(input().split())
set_b = set(input().split())
score = 0

for i in array:
    if i in set_a:
        score += 1
    elif i in set_b:
        score -= 1

print(score)
```

```python
n = int(input())
s = set(map(int, input().split()))
for _ in range(0, int(input())):
    command = input().split()
    match command[0]:
        case 'remove':
            s.remove(int(command[1]))
        case 'discard':
            s.discard(int(command[1]))
        case 'pop':
            s.pop()
print(sum(s))
    
```

```python
_ = input()
a = set(map(int, input().split()))
n = int(input())

for _ in range(n):
    operation = input().split()[0]
    other_set = set(map(int, input().split()))
    getattr(a, operation)(other_set)

print(sum(a))
```

```python
k = int(input())
rooms = list(map(int, input().split()))


captain_room = (sum(set(rooms)) * k - sum(rooms)) // (k - 1)

print(captain_room)
```

```python
a = set(map(int, input().split()))
n = int(input())

is_strict_superset = True

for _ in range(n):
    other = set(map(int, input().split()))
    if not (a > other):
        is_strict_superset = False
        break

print(is_strict_superset)
```
