```python
students = []
scores = set()
for _ in range(int(input())):
        name = input()
        score = float(input())
        scores.add(score)
        students.append([name, score])

scores = list(sorted(scores))
students = sorted(students)

for student in students:
        if student[1] == scores[1]:
                print(student[0])
```

```python
n = int(input())
student_marks = {}
for _ in range(n):
    name, *line = input().split()
    scores = list(map(float, line))
    student_marks[name] = scores
query_name = input()

grades = student_marks[query_name]
ans = sum(grades)/len(grades)
print(f"{ans:.2f}")
```

```python
if __name__ == '__main__':
    N = int(input())
    res = []
    
    for _ in range(N):
        cmd = input().split()
        op = cmd[0]
        
        if op == 'insert':
            res.insert(int(cmd[1]), int(cmd[2]))
        elif op == 'print':
            print(res)
        elif op == 'remove':
            res.remove(int(cmd[1]))
        elif op == 'append':
            res.append(int(cmd[1]))
        elif op == 'sort':
            res.sort()
        elif op == 'pop':
            res.pop()
        elif op == 'reverse':
            res.reverse()
```
