```python
def count_substring(string, sub_string):
index_appear = set()
for i in range(0,len(string)):
    index_appear.add(string.find(sub_string,i))
return len(index_appear.difference({-1}))


if __name__ == '__main__':
string = input().strip()
sub_string = input().strip()

count = count_substring(string, sub_string)
print(count)
```

```python
def wrap(string, max_width):
    l = len(string)
    count=0
    for i in range(1, (l//max_width)+1):
        string = string[:(max_width*i)+count] + '\n' + string[(max_width*i)+count:]
        count+=1
    return string
```

```python
def solve(s):
    name = ''
    for i in range(0,len(s)):
        if s[i-1] == ' ' or i == 0:
            name += s[i].upper()
        else: 
            name += s[i].lower()
    return name
```

```python
def minion_game(string):
    vowel = {'A','E','I','O','U'}
    kevin = 0
    stuart = 0
    lenth = len(string)
    for i in range(0,lenth):
        if string[i] in vowel:
            kevin += lenth - i
        if string[i] not in vowel:
            stuart += lenth - i
    
    if kevin > stuart:
        print('Kevin ' + str(kevin))
    elif kevin == stuart:
        print('Draw')
    elif kevin < stuart:
        print('Stuart ' + str(stuart))

if __name__ == '__main__':
    s = input()
    minion_game(s)
```
