```python
import re

n = int(input())

pattern = r"#([a-fA-F0-9]{6}|[a-fA-F0-9]{3})(?![a-fA-F0-9])"

in_block = False

for _ in range(n):
    line = input()
    
    if "{" in line:
        in_block = True
    elif "}" in line:
        in_block = False
    elif in_block:
        matches = re.findall(pattern, line)
        for m in matches:
            print(f"#{m}")
```

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(f"Start : {tag}")
        for attr, value in attrs:
            print(f"-> {attr} > {value}")

    def handle_endtag(self, tag):
        print(f"End   : {tag}")

    def handle_startendtag(self, tag, attrs):
        print(f"Empty : {tag}")
        for attr, value in attrs:
            print(f"-> {attr} > {value}")

parser = MyHTMLParser()

n = int(input())
for _ in range(n):
    parser.feed(input())
```

```python
from html.parser import HTMLParser


class MyHTMLParser(HTMLParser):

  def handle_comment(self, comment):
    if "\n" in comment:
      print(">>> Multi-line Comment")
    else:
      print(">>> Single-line Comment")
    print(comment)

  def handle_data(self, data):
    if data == "\n":
      return
    print(">>> Data")
    print(data)


html = ""
for i in range(int(input())):
  html += input().rstrip() + "\n"

parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_comment(self, data):
        if '\n' in data:
            print(">>> Multi-line Comment")
        else:
            print(">>> Single-line Comment")
        print(data)

    def handle_data(self, data):
        if data != '\n':
            print(">>> Data")
            print(data)

html = ""
for i in range(int(input())):
    html += input().rstrip()
    html += '\n'

parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
import re


pattern = re.compile(
    r"^(?!.*(.).*\1)(?=(?:.*[A-Z]){2})(?=(?:\D*\d){3})[a-zA-Z0-9]{10}$"
)

for _ in range(int(input())):
  uid = input().strip()
  if pattern.match(uid):
    print("Valid")
  else:
    print("Invalid")
```

```python
import re



pattern = re.compile(r"^[456]\d{3}(-?\d{4}){3}$")

for _ in range(int(input())):
  card = input().strip()

  # Check basic format with regex
  if not pattern.match(card):
    print("Invalid")
    continue

  # Remove hyphens to check consecutive repetitions and exact digit count
  clean_card = card.replace("-", "")

  # Check for 4 or more consecutive repeated digits (e.g., 4444)
  if re.search(r"(\d)\1{3}", clean_card):
    print("Invalid")
  else:
    print("Valid")
```
