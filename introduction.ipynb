```python
n = int(input())
if n % 2 != 0: 
    print("Weird") 
elif 2 <= n <= 5: 
    print("Not Weird") 
elif 6 <= n <= 20: 
    print("Weird") 
else: # n > 20
    print("Not Weird")

```

```python
def calculator(a,b):
    if (1 <= a <= 10**10) and (1 <= b <= 10**10):
        print(a+b)
        print(a-b)
        print(a*b)
    else:
        print("'a' and/or 'b' not valid")

if __name__ == '__main__':
    a = int(input())
    b = int(input())
    calculator(a,b)
```

```python
def __init__(self, integer, floating):
    self.integer = integer
    self.floating = float(floating)


def printtotal(self):
    integer1 = self.integer // self.floating
    integer2 = self.integer / self.floating
    print(f"{integer1}")
    print(f"{integer2}")
```

```python
def function(n):
    positive_int = [x for x in range(0,n)]
    
    for x in positive_int:
        print(x**2)

if __name__ == '__main__':
    n = int(input())
    function(n)
```

```python
def is_leap(year):
    leap = False
    if year%4==0 and year%100!=0:
        leap=True
    elif year%400==0:
        leap=True
    return leap
```
